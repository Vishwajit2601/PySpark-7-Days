```python
%python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("PySpark_Day1") \
    .getOrCreate()

print("Spark Version:", spark.version)
```

```python
data = [
    (101, "Amit", "Pune", 50000),
    (102, "Rahul", "Mumbai", 60000),
    (103, "Priya", "Delhi", 70000),
    (104, "Sneha", "Pune", 55000),
    (105, "Rohit", "Mumbai", 65000)
]

df = spark.createDataFrame(
    data,
    ["customer_id", "name", "city", "salary"]
)

df.show()
```

```python
df.printSchema()
```

```python
df.select(
    "customer_id",
    "name"
).show()
```

```python
df.filter(
    df.salary > 60000
).show()
```

```python
df.count()
```

### Transformation vs Action

```python
'''
# Transformation:-
# A transformation creates a new DataFrame/RDD.
Examples:

select()
filter()
where()
withColumn()
join()
groupBy()
'''

df2 = df.filter(df.salary > 60000)
```

```python
'''
Action:-
An action triggers execution and produces a result/output.

Examples:
    show()
    count()
    collect()
    first()
    take()
    write
'''
df2.show()
```

### Lazy Evaluation

```python
df2 = df.filter(df.salary > 60000)

df3 = df2.select("name")

# Spark doesn't immediately process all the data just because you wrote these transformations.
```

```python
# Spark waits until an action occurs:

df3.show()
# Then execution is triggered.

# Interview answer
# Lazy evaluation means Spark does not immediately execute transformations. It builds an execution plan and executes the required transformations when an action is called.
```

### DAG = Directed Acyclic Graph

### Spark builds a logical execution graph from your operations

```python
df2 = df.filter(df.salary > 60000)

df3 = df2.select("name")

df3.show()
```

### Day-1 Task


```python
data =  [
    (1, "Vishwajit", "IT", 60000),
    (2, "Vikrant", "HR", 45000),
    (3, "Krishna", "IT", 75000),
    (4, "Tanmay", "Finance", 55000),
    (5, "Deepak", "IT", 80000)
]

column = ["emp_id", "name", "dept", "salary"]

df = spark.createDataFrame(data, column)
```

```python
# Display the DataFrame
df.display()

# Display the schema.
df.printSchema()

# Display only name, department,salary
df.select("name", "dept", "salary").display()

# Find employees with salary > 60,000.
df.filter(df.salary > 60000).display()

# Find employees from IT.
df.filter(df.dept == "IT").display()

# Count total employees.
df.count()

```

```python
# Find employees from IT whose salary is greater than 60,000.
df.filter((df.dept == "IT") & (df.salary > 60000)).display()
```
